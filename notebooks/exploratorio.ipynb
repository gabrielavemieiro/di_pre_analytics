```python
caminho_pasta = r"C:\Users\gabriela.vemieiro_ip\Desktop\di_pre_analytics\data\raw"
padrao_arquivos = os.path.join(caminho_pasta, "TaxaReferencia_PRE_*.csv")
arquivos = glob.glob(padrao_arquivos)
 
lista_dataframes = []

for arquivo in arquivos:
    
    nome_arquivo = os.path.basename(arquivo)

    # Regex pra buscar a data (Ex: 20260902)
    match = re.search(r"\d{8}", nome_arquivo)

    if match:
        data_str = match.group(0)  # Exemplo: "20260902"
    else:
        data_str = None  # Caso algum arquivo não tenha o padrão de data

    try:
        df = pd.read_csv(arquivo, sep=";", encoding='latin1') 
    except Exception as e:
        print(f"Erro ao ler {nome_arquivo}: {e}")
        continue

    # Adiciona a coluna com a data extraída
    df["Data"] = data_str
    df['Data'] = pd.to_datetime(data_str, format='%Y%m%d')

    lista_dataframes.append(df)

# Consolida todos os df em um único
if lista_dataframes:
    df_consolidado = pd.concat(lista_dataframes, ignore_index=True)

    # Salva o resultado final em um novo arquivo CSV
    arquivo_saida = "TaxaReferencia_PRE_Consolidado_2.csv"
    df_consolidado.to_csv(arquivo_saida, index=False, encoding="utf-8-sig")

    print(
        f"Consolidação concluída com sucesso! Total de linhas: {len(df_consolidado)}"
    )
    print(f"Arquivo salvo como: {arquivo_saida}")
else:
    print("Nenhum arquivo correspondente foi encontrado.")
```
